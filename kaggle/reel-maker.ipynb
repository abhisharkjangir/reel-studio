# Reel Studio · Reel Maker

Save Version → Save & Run All. This picks up any pending reels from your Blob queue and renders them one by one.

**Required Kaggle Secrets:**
- `OPENAI_API_KEY`
- `ELEVENLABS_API_KEY`
- `SARVAM_API_KEY` *(optional — unlocks Tamil/Telugu/etc. and acts as fallback)*
- `BLOB_READ_WRITE_TOKEN`

Internet + GPU T4×2 must be enabled under Settings → Accelerator.

```python
import os, subprocess, sys
from pathlib import Path

# Pull secrets into the env.
from kaggle_secrets import UserSecretsClient
u = UserSecretsClient()
for name in ("OPENAI_API_KEY", "ELEVENLABS_API_KEY", "SARVAM_API_KEY", "BLOB_READ_WRITE_TOKEN"):
    try:
        os.environ[name] = u.get_secret(name)
    except Exception:
        if name not in ("SARVAM_API_KEY",):
            print(f"WARN · missing secret {name}")

# Clone/update the worker code from GitHub so you only update in one place.
# Replace this URL with your own fork after you push the repo.
REPO = os.environ.get("REEL_STUDIO_REPO", "https://github.com/YOU/reel-studio.git")
root = Path("/kaggle/working/reel-studio")
if not root.exists():
    subprocess.run(["git", "clone", "--depth", "1", REPO, str(root)], check=True)
sys.path.insert(0, str(root / "kaggle"))

subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r", str(root / "kaggle" / "requirements.txt")], check=True)
```
